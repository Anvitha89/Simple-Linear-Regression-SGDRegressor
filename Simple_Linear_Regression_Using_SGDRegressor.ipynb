```python
import numpy as np
import pandas as pd
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error
```

```python
X = 2* np.random.rand(1000, 1)
y = 4 * X + 3
```

```python
#train the model
```

```python
sgd_reg = SGDRegressor(max_iter = 1000, tol = 0.01, eta0 = 0.01, random_state = 76)
sgd_reg.fit(X, y.ravel())
```

```python
sgd_reg.coef_, sgd_reg.intercept_
```

```python
sgd_reg.predict([[1]])
```

```python
y_pred = sgd_reg.predict(X)
print(y_pred)
```

```python
mse = mean_squared_error(y.ravel(), y_pred)
print(mse)
```

```python
rmse = np.sqrt(mse)
print('rmse:', rmse)
```
